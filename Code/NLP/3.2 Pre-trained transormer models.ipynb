# 4.3 Pre-trained Transformer Models

```python
import transformers
from transformers import pipeline
import torch
```

```python
sentiment_pipeline = pipeline("sentiment-analysis", device='mps')
```

```python
sentence_1 = "i had a great time at the movie it was really funny"
sentence_2 = "i had a great time at the movie but the parking was terrible"
sentence_3 = "i had a great time at the movie but the parking wasn't great"
sentence_3_bis = "i had a great time at the movie but the parking was not great"
sentence_4 = "i went to see a movie"
sentence_5 = "the movie was terrible"
```

```python
print(sentence_1)
sentiment_pipeline(sentence_1)
```

```python
test = sentiment_pipeline([sentence_1, sentence_2])
[sub['label'] for sub in test]
```

```python
print(sentence_2)
sentiment_pipeline(sentence_2)
```

```python
print(sentence_3)
sentiment_pipeline(sentence_3)
```

```python
print(sentence_4)
sentiment_pipeline(sentence_4)
```

```python
print(sentence_5)
sentiment_pipeline(sentence_5)
```

The default model isn't giving us great results for our neutral sentence. However there are loads of models we can choose to use that have been trained on different data with different parameters: https://huggingface.co/models?pipeline_tag=text-classification&sort=downloads&search=sentiment

Let's try a specific model trained on tweets in the English language only and see if that improves our results.

```python
specific_model = pipeline("sentiment-analysis", model="finiteautomata/bertweet-base-sentiment-analysis")
```

```python
print(sentence_1)
specific_model(sentence_1)
```

```python
print(sentence_2)
specific_model(sentence_2)
```

```python
print(sentence_3)
specific_model(sentence_3)
```

```python
print(sentence_4)
specific_model(sentence_4)
```

```python
print(sentence_5)
specific_model(sentence_5)
```

## Italian Sentiment Analysis

**MilaNLProc/feel-it-italian-sentiment**: è uno dei modelli più diffusi per il sentiment in italiano, basato su AlBERTo e addestrato su dataset annotati manualmente

```python

# Modello per Sentiment Analysis in italiano
sentiment_it = pipeline(
    "sentiment-analysis",
    model="MilaNLProc/feel-it-italian-sentiment",
    tokenizer="MilaNLProc/feel-it-italian-sentiment"
)
```

```python
sentiment_it("Questo corso di NLP faceva letterlamente schifo")
```

```python
sentiment_it("Questo corso di NLP é stato come un bagno nell'acido")
```

```python
sentiment_it("Il docente era veramente fantastico ")
```

```python
sentiment_it("Il corso era carino")
```
