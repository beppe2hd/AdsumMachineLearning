# 2.6 Stemming

Stemming: si riferisce a standardizzare il testo, ossia la riduzione delle parole alla loro forma base. Ad esempio, parole come “connecting” o ‘connected’ saranno ridotte alla forma base “connect”. Lo stemming funziona rimuovendo il suffisso/la desinenza della parola, ma a volte può portare alla formazione di una forma base priva di significato o non corretta.

Standardizziamo il testo in questo modo perché riduce il numero di parole uniche nel nostro set di dati, riducendo così le dimensioni e la complessità dei nostri dati. 

```python
from nltk.stem import PorterStemmer
```

```python
# create stemmer
ps = PorterStemmer()
```

```python
connect_tokens = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

```python
for t in connect_tokens:
    print(t, " : ", ps.stem(t))
```

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

```python
for t in learn_tokens:
    print(t, " : ", ps.stem(t))
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

```python
for t in likes_tokens:
    print(t, " : ", ps.stem(t))
```
