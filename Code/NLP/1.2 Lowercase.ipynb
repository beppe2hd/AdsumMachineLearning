# 2.2 Lowercase

Un primo passo importante nel lavorare con i dati testuali è semplicemente convertirli in minuscolo. 
Tale operazione aiuta a mantenere la coerenza sia nei dati che nei risultati. Quando lavoriamo con il testo, che si tratti di analisi esplorativa o di machine learning, vogliamo assicurarci che le parole siano comprese e conteggiate come la stessa parola. Infatti u modello di analisi potrebbe trattare una parola con una lettera maiuscola in modo diverso dalla stessa parola senza alcuna lettera maiuscola. La conversione in minuscolo garantisce la conformità e rende più facile continuare con un'ulteriore pulizia dei dati.

Tuttavia, è opportuno ricordate che la conversione in minuscolo può cambiare il significato di alcuni testi, ad esempio “US” in maiuscolo è inteso come un paese, al contrario di “us”. 

```python
sentence = "Her cat's name is Luna"
print(sentence)
```

```python
lower_sentence = sentence.lower()
print(lower_sentence)
```

```python
sentence_list = ['Could you pass me the TV remote?', 
                 'It is IMPOSSIBLE to find this hotel', 
                 'Want to go for dinner on Tuesday?']
print(sentence_list)
```

```python
lower_sentence_list = [x.lower() for x in sentence_list]
print(lower_sentence_list)
```
