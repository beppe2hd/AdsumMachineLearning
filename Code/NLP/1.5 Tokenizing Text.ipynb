# 2.5 Tokenizing Text

Un passo fondamentale nell'NLP consiste nel convertire il nostro testo in unità più piccole attraverso un processo noto come **tokenizzazione**. Queste unità più piccole sono note come token. La tokenizzazione delle parole è la forma più comune di tokenizzazione, in cui le singole parole del testo diventano un token, ma i token possono anche essere frasi, sottoword o singoli caratteri a seconda del caso d'uso. 

Perché lo facciamo? Il significato del testo nel suo complesso risulta più chiaro se riusciamo ad analizzare e comprendere sia le singole parti che l'insieme.

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

### Sentance tokenization

```python
sentences = "Her cat's name is Luna. Her dog's name is max"
sent_tokenize(sentences)
```

### Word tokenization

```python
sentence = "Her cat's name is Luna"
word_tokenize(sentence)
```

Notare come "cat's"'è stato diviso in 2 tokens.

```python
sentence_2 = "Her cat's name is Luna and her dog's name is max"
word_tokenize(sentence_2)
```

Possiamo vedere che abbiamo due occorrenze della parola “her”, una delle quali è scritta con l'iniziale maiuscola. I token sono quindi diversi e saranno trattati come tali nella maggior parte delle analisi.
