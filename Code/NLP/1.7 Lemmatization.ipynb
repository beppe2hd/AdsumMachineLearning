# 2.7 Lemmatization

Lemmatization riduce la parola alla sua forma base più significativa e garantisce che non perda il suo significato. La lemmatizzazione funziona in modo più intelligente, facendo riferimento a un dizionario predefinito contenente il contesto delle parole e utilizzandolo quando riduce la parola alla sua forma base.

```python
connect_tokens = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

## Stemming

```python
from nltk.stem import PorterStemmer
```

```python
# create stemmer
ps = PorterStemmer()
```

```python
for t in connect_tokens:
    print(t, " : ", ps.stem(t))
```

```python
for t in learn_tokens:
    print(t, " : ", ps.stem(t))
```

```python
for t in likes_tokens:
    print(t, " : ", ps.stem(t))
```

## Lemmatization

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
# create lemmatizer 
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in learn_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in likes_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```
